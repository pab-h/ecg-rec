```python
import pandas as pd
```

---

```python
leads = ["LI", "LII", "LIII", "aVR", "aVL", "aVF", "V1", "V2", "V3", "V4", "V5", "V6"]
```

---

```python
exams = pd.read_csv("../data/exams.csv")

exams.head()
```

---

```python
exams_testset = exams[
    exams["trace_file"] == "exams_part3.hdf5"
]

exams_testset.shape
```

```python
exams_testset = exams_testset.drop_duplicates("patient_id", keep = "first")

exams_testset.shape
```

---

Normal

```python
exams_testset_normal = exams_testset[exams_testset["normal_ecg"] == True]
```

```python
exams_testset_normal.shape
```

```python
exams_testset_normal["category"] = "normal"
```

1dAVb

```python
exams_testset_1dAVb = exams_testset[
    (exams_testset["1dAVb"] == True) &
    (exams_testset["RBBB"] == False) &
    (exams_testset["LBBB"] == False) &
    (exams_testset["SB"] == False)   &
    (exams_testset["ST"] == False)   &
    (exams_testset["AF"] == False)        
]
```

```python
exams_testset_1dAVb.shape
```

```python
exams_testset_1dAVb["category"] = "1dAVb"
```

RBBB

```python
exams_testset_RBBB = exams_testset[
    (exams_testset["1dAVb"] == False) &
    (exams_testset["RBBB"] == True)   &
    (exams_testset["LBBB"] == False)  &
    (exams_testset["SB"] == False)    &
    (exams_testset["ST"] == False)    &
    (exams_testset["AF"] == False)        
]
```

```python
exams_testset_RBBB.shape
```

```python
exams_testset_RBBB["category"] = "RBBB"
```

LBBB

```python
exams_testset_LBBB = exams_testset[
    (exams_testset["1dAVb"] == False) &
    (exams_testset["RBBB"] == False)  &
    (exams_testset["LBBB"] == True)   &
    (exams_testset["SB"] == False)    &
    (exams_testset["ST"] == False)    &
    (exams_testset["AF"] == False)        
]
```

```python
exams_testset_LBBB.shape
```

```python
exams_testset_LBBB["category"] = "LBBB"
```

SB

```python
exams_testset_SB = exams_testset[
    (exams_testset["1dAVb"] == False) &
    (exams_testset["RBBB"] == False)  &
    (exams_testset["LBBB"] == False)  &
    (exams_testset["SB"] == True)     &
    (exams_testset["ST"] == False)    &
    (exams_testset["AF"] == False)        
]
```

```python
exams_testset_SB.shape
```

```python
exams_testset_SB["category"] = "SB"
```

ST

```python
exams_testset_ST = exams_testset[
    (exams_testset["1dAVb"] == False) &
    (exams_testset["RBBB"] == False)  &
    (exams_testset["LBBB"] == False)  &
    (exams_testset["SB"] == False)    &
    (exams_testset["ST"] == True)     &
    (exams_testset["AF"] == False)        
]
```

```python
exams_testset_ST.shape
```

```python
exams_testset_ST["category"] = "ST"
```

AF

```python
exams_testset_AF = exams_testset[
    (exams_testset["1dAVb"] == False) &
    (exams_testset["RBBB"] == False)  &
    (exams_testset["LBBB"] == False)  &
    (exams_testset["SB"] == False)    &
    (exams_testset["ST"] == False)    &
    (exams_testset["AF"] == True)        
]
```

```python
exams_testset_AF.shape
```

```python
exams_testset_AF["category"] = "AF"
```

```python
exams_testset_exclusive_report = pd.concat([
    exams_testset_1dAVb, 
    exams_testset_RBBB, 
    exams_testset_LBBB, 
    exams_testset_SB, 
    exams_testset_ST, 
    exams_testset_AF,
    exams_testset_normal
])
```

```python
exams_testset_exclusive_report.shape
```

---

```python
mae_scores = pd.read_csv("../dist/mae.scores.csv")
```

```python
mae_scores_exclusive_report = pd.merge(
    mae_scores, 
    exams_testset_exclusive_report[["exam_id", "category"]], 
    on  = "exam_id", 
    how = 'inner'
)

mae_scores_exclusive_report.head()
```

```python
mae_scores_exclusive_report.groupby("category").mean()[leads]
```

---

```python
r2_scores = pd.read_csv("../dist/r2.scores.csv")
```

```python
r2_scores_exclusive_report = pd.merge(
    r2_scores, 
    exams_testset_exclusive_report[["exam_id", "category"]], 
    on  = "exam_id", 
    how = 'inner'
)

r2_scores_exclusive_report.head()
```

```python
r2_scores_exclusive_report.groupby("category").mean()[leads]
```

```python
corr_scores_exclusive_report = pd.merge(
    corr_scores, 
    exams_testset_exclusive_report[["exam_id", "category"]], 
    on  = "exam_id", 
    how = 'inner'
)

corr_scores_exclusive_report.head()
```
