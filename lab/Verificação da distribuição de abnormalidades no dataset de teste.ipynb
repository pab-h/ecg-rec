Verificação da distribuição de abnormalidades no dataset de teste

---

Bibliotecas

```python
import pandas as pd
```

---

Carregando o dataset de teste

```python
exams = pd.read_csv("../data/exams.csv")

exams.head()
```

```python
exams_testset = exams[
    exams["trace_file"] == "exams_part3.hdf5"
]

exams_testset.shape
```

```python
exams_testset = exams_testset.drop_duplicates(
    subset       = "patient_id", 
    keep         = "first", 
    ignore_index = True
)

exams_testset.shape
```

---

Verificando a distribuição de ecgs normais/anormais

```python
exams_testset["normal_ecg"].value_counts()
```

---
