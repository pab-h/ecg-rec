Geração dos resultados sem repetição de paciente

---

```python
import pandas as pd
```

```python
exams = pd.read_csv("../data/exams.csv")

exams.head()
```

```python
exams_patient = exams[
    exams["trace_file"] == "exams_part3.hdf5"
][["exam_id", "patient_id"]]

exams_patient.head()
```

```python
unique_exams_patient = exams_patient.drop_duplicates(subset=["patient_id"])

unique_exam_ids = unique_exams_patient["exam_id"]

unique_exam_ids.shape
```

---

```python
corr_dataset = pd.read_csv("../dist/corr.scores.csv")
corr_dataset = corr_dataset.set_index("exam_id")

corr_dataset.shape
```

```python
corr_dataset_filtered = corr_dataset.loc[corr_dataset.index.isin(unique_exam_ids)]

corr_dataset_filtered.shape
```

```python
corr_dataset_filtered.to_csv("../dist/corr.scores.unique-patient.csv")
```
