# Verificando a estratificação do dataset

## Bibliotecas

```python
import pandas as pd
```

## Configurações

```python
EXAMS_CSV = "../data/exams.csv"

COLUMNS = [
    "normal_ecg",
    "is_male",
    "1dAVb",
    "RBBB",
    "LBBB",
    "SB",
    "ST",
    "AF",
]
```

## Verificação

```python

exams = pd.read_csv(EXAMS_CSV)

results = []

for trace_file, subset in exams.groupby("trace_file"):

    counts = {
        "trace_file": trace_file,
        "num_exams":  len(subset)
    }

    for col in COLUMNS:
        counts[col] = subset[col].fillna(False).astype(bool).sum()

    results.append(counts)

summary = pd.DataFrame(results)

print(summary)
```

## Salvando os resultados

```python
summary.to_csv("../dist/dataset-summary.csv", index = False)
```
