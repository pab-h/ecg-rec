Análise dos resultados agregados por número de derivações usadas

---

```python
import pandas as pd
import numpy  as np
```

---

```python
corr_df = pd.read_csv("../dist/corr.scores.agg.csv", index_col = "n_derivations")
mae_df  = pd.read_csv("../dist/mae.scores.agg.csv", index_col = "n_derivations")
r2_df   = pd.read_csv("../dist/r2.scores.agg.csv", index_col = "n_derivations")
```

```python
ecg_colums = ["LI", "LII", "LIII", "aVR", "aVL", "aVF", "V1", "V2", "V3", "V4", "V5", "V6"]

corr_df = corr_df.loc[:, ecg_colums]
mae_df  = mae_df.loc[:, ecg_colums]
r2_df   = r2_df.loc[:, ecg_colums]
```

---

```python
distinct_colors = [
    "#e6194B",  # Vermelho
    "#3cb44b",  # Verde
    "#ffe119",  # Amarelo
    "#4363d8",  # Azul
    "#f58231",  # Laranja
    "#911eb4",  # Roxo
    "#42d4f4",  # Ciano
    "#f032e6",  # Magenta
    "#bfef45",  # Verde-lima
    "#fabed4",  # Rosa
    "#469990",  # Teal / Verde-azulado
    "#9A6324",  # Marrom
]
```

---

```python
np.round(corr_df, 2)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(11, 6), dpi=300)

for i, col in enumerate(corr_df.columns):
    plt.plot(corr_df.index, corr_df[col], marker="o", label=col, alpha=0.85, color = distinct_colors[i])

plt.plot(
    corr_df.index,
    corr_df.mean(axis=1),
    color="black",
    linestyle="--",
    linewidth=2.5,
    label="Overall Average",
)

plt.title(
    "Evolution of Correlation by Number of Leads",
    fontsize=13,
    fontweight="bold",
)
plt.xlabel("Number of Leads Used", fontsize=11)
plt.ylabel("Correlation", fontsize=11)
plt.grid(True, linestyle=":", alpha=0.6)
plt.legend(
    bbox_to_anchor=(1.02, 1), loc="upper left", frameon=True, title="Lead"
)
plt.tight_layout()
plt.show()
```

---

```python
import matplotlib.pyplot as plt

metrics = [
    (mae_df,  "Evolution of MAE",         "MAE"),
    (r2_df,   "Evolution of R²",          "$R^2$"),
    (corr_df, "Evolution of Correlation", "Correlation"),
]

fig, axes = plt.subplots(1, 3, figsize=(18, 6), dpi=300)

for ax, (df, title, ylabel) in zip(axes, metrics):
    for i, col in enumerate(df.columns):
        ax.plot(
            df.index,
            df[col],
            marker="o",
            label=col,
            alpha=0.85,
            color=distinct_colors[i],
        )

    ax.plot(
        df.index,
        df.mean(axis=1),
        color="black",
        linestyle="--",
        linewidth=2.5,
        label="Overall Average",
    )

    ax.set_title(title, fontsize=20, fontweight="bold")
    ax.set_ylabel(ylabel, fontsize=20)
    ax.grid(True, linestyle=":", alpha=0.6)

axes[-2].set_xlabel("Number of leads used for reconstruction", fontsize=20)

handles, labels = axes[0].get_legend_handles_labels()
fig.legend(
    handles,
    labels,
    bbox_to_anchor=(1.01, 0.5),
    loc="center left",
    frameon=True,
    fontsize=20
    
)

plt.tight_layout()
plt.show()
```
